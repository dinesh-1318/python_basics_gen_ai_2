### **pandas series**

```python
import pandas as pd
```

```python
name = pd.Series(["Dinesh","Harish","Ganesh"])
name
```

```python
age = pd.Series([21,23,20])
age
```

#### **key setting in series**

```python
car = pd.Series(["benz","bmw","tata"],index=["A","B","C"])
car
```

### **DataFrame in pandas**

```python
student = pd.DataFrame({
    "st_name":name,
    "st_age":age
})
student
```
